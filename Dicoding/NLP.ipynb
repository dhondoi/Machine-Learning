## **Tokenizer dan Sequence**
Proses mengonversi kata-kata ke dalam bilangan numerik dapat kita sebut juga sebagai tokenisasi. Setelah melakukan tokenisasi pada teks, hal selanjutnya adalah mengubah setiap kalimat dalam teks kita ke dalam sequence. Sebuah sekuens adalah sebuah larik yang berisi kumpulan token sesuai dengan setiap kata pada sebuah kalimat dalam teks.

```python
# import library
from tensorflow.keras.preprocessing.text import Tokenizer
"""
objek tokenizer dengan memanggil fungsi tokenizer dan melengkapi parameternya.
Parameter num_words adalah jumlah kata yang akan dikonversi ke dalam
token/bilangan numerik. Jika parameter num_words diatur sebanyak 15, hanya 15
kata yang paling sering muncul. 15 kata tersebut akan ditokenisasi dari seluruh
kata pada dataset.

Sedangkan parameter oov_token adalah parameter yang berfungsi untuk mengganti
kata-kata yang tidak ditokenisasi menjadi karakter tertentu.
Pada praktiknya, lebih baik untuk mengganti kata yang tidak dikenali dengan
suatu kata tertentu dibanding melewatkan kata tersebut untuk mengurangi
informasi yang hilang. Hal inilah yang dapat dilakukan dengan
menambahkan parameter OOV.
"""
tokenizer = Tokenizer(num_words= 15, oov_token='-')
# teks yang akan kita tokenisasi dan kita pakai untuk pelatihan model.
teks = ['Saya suka programming.',
        'Programming sangat menyenangkan!',
        'Machine Learning berbeda dengan pemrograman konvensional',
        'Haha ngajak ribut kah',
        'ayo aja w mah',
        "Think and wonder, wonder and think."
        ]
"""
Melakukan tokenisasi, panggil fungsi fit_on_text() pada objek tokenizer dan
isi teks kita sebagai argumennya.
"""
tokenizer.fit_on_texts(teks)
"""
Mengubah text yang telah dibuat sebelumnya ke dalam bentuk sequence
menggunakan fungsi text_to_sequences.
"""
sequences = tokenizer.texts_to_sequences(teks)
print(sequences)
print(tokenizer.word_index)
"""
padding adalah proses untuk membuat setiap kalimat pada teks memiliki
panjang yang seragam. Sama seperti melakukan resize gambar, agar resolusi
setiap gambar sama besar.
"""
from tensorflow.keras.preprocessing.sequence import pad_sequences
sequences_samapanjang = pad_sequences(sequences)
print("Sama Panjang 1")
print(sequences_samapanjang)
"""
jikalau 0 ingin ditambahkan di akhir sequence, kita dapat menggunakan
parameter padding dengan nilai ‘post’. Selain itu kita dapat mengatur
parameter maxlen (panjang maksimum setiap sequence) dengan nilai yang
kita inginkan. Jika kita mengisi nilai 5, maka panjang sebuah
sequence tidak akan melebihi 5.
"""
sequences_samapanjang2 = pad_sequences(sequences,
                                      padding='post',
                                      maxlen=5)
print("Sama Panjang 2")
print(sequences_samapanjang2)
"""
Untuk mengubah pengaturan ini dan mengambil 5 kata awal dari tiap kalimat,
kita dapat menggunakan parameter truncating dan mengisi nilai ‘post’.
"""
print("Sama Panjang 3")
sequences_samapanjang3 = pad_sequences(sequences,
                                      padding='post',
                                      maxlen=5,
                                      truncating='post')
print(sequences_samapanjang3)
```

## **Practice 1 : Binary Text Classification**


```python
# Data Collecting
import pandas as pd

df = pd.read_csv('yelp_labelled.txt', names=['sentence', 'label'], sep='\t')
# print(df)

# Data Preparation
from sklearn.model_selection import train_test_split

kalimat = df['sentence'].values
print(kalimat.shape)

y = df['label'].values
print(y.shape)

kalimat_latih, kalimat_test, y_latih, y_test = train_test_split(kalimat, y, test_size=0.2)

# Tokenizer (Data Cleaning)
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences

tokenizer = Tokenizer(num_words=250, oov_token='x')
tokenizer.fit_on_texts(kalimat_latih)
# print(tokenizer.word_index)
# print(len(tokenizer.word_index))
tokenizer.fit_on_texts(kalimat_test)

sekuens_latih = tokenizer.texts_to_sequences(kalimat_latih)
# print(sekuens_latih)
sekuens_test = tokenizer.texts_to_sequences(kalimat_test)

padded_latih = pad_sequences(sekuens_latih, maxlen=20)
# print(padded_latih[0])
padded_test = pad_sequences(sekuens_test, maxlen=20)

# Modelling
import tensorflow as tf
model = tf.keras.Sequential([
    tf.keras.layers.Embedding(250, 16, input_length=20),
    tf.keras.layers.GlobalAveragePooling1D(),
    tf.keras.layers.Dense(24, activation='relu'),
    tf.keras.layers.Dense(1, activation='sigmoid')
])
model.summary()
model.compile(loss='binary_crossentropy',optimizer='adam',metrics=['accuracy'])

# Fitting Model
num_epochs = 30
history = model.fit(padded_latih, y_latih, epochs=num_epochs,
                    validation_data=(padded_test, y_test), verbose=2)
```

## **Practice 2 : Multiclass Text Classification**


```python
# Data Collecting
import pandas as pd

df = pd.read_csv('imdb_indonesian_movies_2.csv')
df = df.drop(columns=['judul_film'])

category = pd.get_dummies(df.genre)
df_baru = pd.concat([df, category], axis=1)
df_baru = df_baru.drop(columns='genre')
# print(df_baru)

sinopsis = df_baru['ringkasan_sinopsis'].values
# print(sinopsis)
# print(sinopsis.shape)
label = df_baru[['Drama', 'Horor', 'Komedi', 'Laga', 'Romantis']].values
# print(label)
# print(label.shape)

# Data Preparation
from sklearn.model_selection import train_test_split
sinopsis_latih, sinopsis_test, label_latih, label_test = train_test_split(sinopsis, label, test_size=0.2)

# Data Cleaning
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences

tokenizer = Tokenizer(num_words=10000, oov_token='x')
tokenizer.fit_on_texts(sinopsis_latih)
tokenizer.fit_on_texts(sinopsis_test)

sekuens_latih = tokenizer.texts_to_sequences(sinopsis_latih)
sekuens_test = tokenizer.texts_to_sequences(sinopsis_test)

padded_latih = pad_sequences(sekuens_latih)
# print(padded_latih)
padded_test = pad_sequences(sekuens_test)

# Modelling
import tensorflow as tf
model = tf.keras.Sequential([
    tf.keras.layers.Embedding(input_dim=10000, output_dim=16),
    tf.keras.layers.LSTM(64),
    tf.keras.layers.Dense(128, activation='relu'),
    tf.keras.layers.Dense(64, activation='relu'),
    tf.keras.layers.Dense(5, activation='softmax')
])
model.compile(loss='categorical_crossentropy',optimizer='adam',metrics=['accuracy'])

# Fitting Model
num_epochs = 30
history = model.fit(padded_latih, label_latih, epochs=num_epochs,
                    validation_data=(padded_test, label_test), verbose=2)
```

## **Practice 3 : Simple Binary Classification**


```python
# Data Collecting
import pandas as pd
import numpy as np

df = pd.read_csv('yelp_labelled.txt', names=['sentence', 'label'], sep='\t')
# print(df)

# Data Preparation
from sklearn.model_selection import train_test_split

positive = df.iloc[np.where(df["label"]==1)].head(100)
negative = df.iloc[np.where(df["label"]==0)].head(100)
df = pd.concat([positive, negative])
attribute = df["sentence"].values
# print(attribute)

label = df['label'].values
# print(label)

attribute_train, attribute_test, label_train, label_test = train_test_split(attribute, label, test_size=0.3)

# Tokenizer (Data Cleaning)
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences

tokenizer = Tokenizer(num_words=800, oov_token='x')
tokenizer.fit_on_texts(attribute_train)
tokenizer.fit_on_texts(attribute_test)
# print(len(tokenizer.word_index))
# print(tokenizer.word_index)

sekuens_latih = tokenizer.texts_to_sequences(attribute_train)
# print(sekuens_latih)
sekuens_test = tokenizer.texts_to_sequences(attribute_test)

padded_latih = pad_sequences(sekuens_latih)
# print(padded_latih[0])
padded_test = pad_sequences(sekuens_test)

# Modelling
import tensorflow as tf
model = tf.keras.Sequential([
    tf.keras.layers.Embedding(800, 16),
    tf.keras.layers.GlobalAveragePooling1D(),
    tf.keras.layers.Dense(24, activation='relu'),
    tf.keras.layers.Dense(1, activation='sigmoid')
])
model.summary()
model.compile(loss='binary_crossentropy',optimizer='adam',metrics=['accuracy'])

# Fitting Model
num_epochs = 30
history = model.fit(padded_latih, label_train, epochs=num_epochs,
                    validation_data=(padded_test, label_test), verbose=2)


# Predict
predict = tokenizer.texts_to_sequences(["Crust is not good."])
predict = pad_sequences(predict)
result = model.predict(predict)
print(result)
```
