```python
import tensorflow as tf
import numpy as np
```

```python
def windowed_dataset(series, window_size, batch_size, shuffle_buffer):

    # membuat array [num1 ... numN] menjadi [[num1] [...] [numN]]
    dataTrain = tf.expand_dims(series, axis=-1)
    print("membuat array [num1 ... numN] menjadi [[num1] [...] [numN]]")
    print(dataTrain)
    print("-----------------------------------------------------------\n")

    # membuat array [[num1] [...] [numN]] menjadi [[[num1]],[[...]],[[numN]]]
    ds = tf.data.Dataset.from_tensor_slices(series)
    print("membuat array [[num1] [...] [numN]] menjadi [[[num1]],[[...]],[[numN]]]")
    print(list(ds.as_numpy_iterator()))
    print("-----------------------------------------------------------\n")

    # membuat window dengan jumlah isi yang telah ditentukan
    ds = ds.window(window_size + 1, shift=1, drop_remainder=True)
    print("membuat window dengan jumlah isi yang telah ditentukan")
    for window in ds:
      print(list(window.as_numpy_iterator()))
    print("-----------------------------------------------------------\n")

    # membuat array menjadi [[num1] [...] [numN]]
    ds = ds.flat_map(lambda w: w.batch(window_size + 1))
    print("membuat array menjadi [[num1] [...] [numN]]")
    for window in ds:
      print(window)
    print("-----------------------------------------------------------\n")

    # dibuat acak
    ds = ds.shuffle(shuffle_buffer)
    print("dibuat acak")
    for window in ds:
      print(window)
    print("-----------------------------------------------------------\n")

    # mapping
    ds = ds.map(lambda w: (w[:-1], w[-1:]))
    print("mapping")
    for window in ds:
      print(window)
    print("-----------------------------------------------------------\n")

    return ds.batch(batch_size).prefetch(1)
```

```python
series = np.arange(10)

windowedTrainData = windowed_dataset(series, 2, 3, 1)

# hasil windowed batch dataset
print("hasil windowed batch dataset")
for window in windowedTrainData:
  print(window)
print("-----------------------------------------------------------\n")
```
