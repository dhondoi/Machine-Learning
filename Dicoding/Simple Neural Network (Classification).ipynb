# **Membuat Atribut dan Label**

```python
import pandas as pd
df = pd.read_csv('citrus.csv')
df.name[df.name == 'orange'] = 0
df.name[df.name == 'grapefruit'] = 1
dataset = df.values
attribute = dataset[:,1:6]
label = dataset[:,0]
```

# **Normalization**

```python
from sklearn import preprocessing
import numpy as np

min_max_scaler = preprocessing.MinMaxScaler()
attribute_scale = min_max_scaler.fit_transform(attribute)

label = label.astype(np.float32)
```

# **Split**

```python
from sklearn.model_selection import train_test_split
attribute_train, attribute_test, label_train, label_test = train_test_split(attribute_scale, label, test_size=0.3)
print(attribute_train)
print(label_train)
```

# **Membuat Model**

```python
from keras.models import Sequential
from keras.layers import Dense
model = Sequential([
                    Dense(32, activation='relu', input_shape=(5,)),
                    Dense(32, activation='relu'),
                    Dense(1, activation='sigmoid'),])
```

# **Menentukan Optimizer dan Loss untuk Model**

```python
model.compile(optimizer='sgd',
              loss='binary_crossentropy',
              metrics=['accuracy'])
```

# **Masukkan Data Latih untuk Model**

```python
model.fit(attribute_train, label_train, epochs=100,batch_size=67)
```

# **Uji Model**

```python
# elemen pertama adalah loss dan elemen kedua adalah akurasi
loss,accuracy = model.evaluate(attribute_test, label_test)
print(f"Loss : {loss}")
print(f"Accuracy : {accuracy}")
```
